```scala
%spark.pyspark
from pyspark.sql import functions as F
from pyspark.sql.window import Window

users = spark.createDataFrame(
    [("u1", "Berlin"),
     ("u2", "Berlin"),
     ("u3", "Munich"),
     ("u4", "Hamburg")],
    ["user_id", "city"]
)

orders = spark.createDataFrame(
    [("o1", "u1", "p1", 2, 10.0),
     ("o2", "u1", "p2", 1, 30.0),
     ("o3", "u2", "p1", 1, 10.0),
     ("o4", "u2", "p3", 5, 7.0),
     ("o5", "u3", "p2", 3, 30.0),
     ("o6", "u3", "p3", 1, 7.0),
     ("o7", "u4", "p1", 10, 10.0)],
    ["order_id", "user_id", "product_id", "qty", "price"]
)

products = spark.createDataFrame(
    [("p1", "Ring VOLA"),
     ("p2", "Ring POROG"),
     ("p3", "Ring TISHINA")],
    ["product_id", "product_name"]
)

users.show()
orders.show()
products.show()
```

```scala
%spark.pyspark

orders_with_revenue = orders.withColumn(
    "revenue", F.col("qty") * F.col("price")
)

mart = (
    orders_with_revenue
    .join(users, on="user_id", how="left")
    .join(products, on="product_id", how="left")
)

mart.show()
```

```scala
%spark.pyspark
agg_mart = mart.groupBy("city", "product_id", "product_name").agg(
    F.count("order_id").alias("orders_cnt"),
    F.sum("qty").alias("qty_sum"),
    F.sum("revenue").alias("revenue_sum")
)

agg_mart.show()
```

```scala
%spark.pyspark
window = Window.partitionBy("city").orderBy(F.desc("revenue_sum"))

mart_city_top_products = (
    agg_mart
    .withColumn("rank", F.rank().over(window))
    .filter(F.col("rank") <= 2)
    .drop("rank")
)

mart_city_top_products.show()
```

```scala
%spark.pyspark
hdfs_path = "/tmp/sandbox_zeppelin/mart_city_top_products/"
s3_path = "s3a://dataproc-bucket/mart_city_top_products/"

mart_city_top_products.write.mode("overwrite").parquet(hdfs_path)
mart_city_top_products.write.mode("overwrite").parquet(s3_path)
```

```scala
%spark.pyspark
result_hdfs = spark.read.parquet(hdfs_path)
print("HDFS")
result_hdfs.orderBy("city", F.desc("revenue_sum")).show()

result_s3 = spark.read.parquet(s3_path)
print ("S3")
result_s3.orderBy("city", F.desc("revenue_sum")).show()
```
